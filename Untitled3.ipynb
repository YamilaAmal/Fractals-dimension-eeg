```python
import mne

# Cargar el archivo EEG de reposo del sujeto 00
rest_raw = mne.io.read_raw_edf("eegmat/Subject00_1.edf", preload=True)

# Mostrar los nombres de los canales
print("Canales disponibles:", rest_raw.ch_names)
```

```python
# Cortamos una copia de los primeros 60 segundos
rest_first_minute = rest_raw.copy().crop(tmin=0, tmax=60)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Extraemos datos y tiempos
data, times = rest_first_minute.get_data(return_times=True)  # data: canales x muestras

# Número de canales y muestras
n_channels, n_samples = data.shape

# Crear figura y eje
fig, ax = plt.subplots(n_channels, 1, figsize=(20, 2*n_channels), sharex=True)

# Para separar las señales verticalmente y evitar superposición
offset = 100  # Ajusta según amplitud de la señal

for i in range(n_channels):
    ax[i].plot(times, data[i] + i*offset, color='black')  # offset para separar canales
    ax[i].set_ylabel(rest_first_minute.ch_names[i], rotation=0, labelpad=40, fontsize=10)
    ax[i].set_yticks([])
    ax[i].spines['top'].set_visible(False)
    ax[i].spines['right'].set_visible(False)
    ax[i].spines['left'].set_visible(False)

ax[-1].set_xlabel("Tiempo (s)")
plt.suptitle("EEG Estado Reposo - Primeros 60 segundos (todos los canales)")
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
import numpy as np

def higuchi_fd(signal, kmax):
    N = len(signal)
    L = []
    x = signal
    for k in range(1, kmax+1):
        Lk = []
        for m in range(k):
            Lmk = 0
            n_max = int(np.floor((N - m - 1) / k))
            for i in range(1, n_max):
                Lmk += abs(x[m + i * k] - x[m + (i - 1) * k])
            Lmk = (Lmk * (N - 1)) / (k * n_max * k)
            Lk.append(Lmk)
        L.append(np.mean(Lk))
    lnL = np.log(L)
    lnk = np.log(1.0 / np.arange(1, kmax + 1))
    return np.polyfit(lnk, lnL, 1)[0]
```
